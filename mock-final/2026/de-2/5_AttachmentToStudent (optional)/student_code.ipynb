# Spark Exam : E-Commerce Business Transcation 
Using the data file in parquet format, analyze the statistics of Sales transaction of a UK-based e-commerce.

## Initilization
Define all parameters

```python
host = 'local'
app_name = 'E-Commerce Analysis'
source_file_path = 'parquet'
data_file_path = '/tmp/ecommerce'
```

Dataset preparation

```python
import shutil

shutil.copytree(source_file_path, data_file_path, dirs_exist_ok=True)
```

Perform all necessary initialization

```python
import os
# This MUST come before 'from pyspark.sql import SparkSession'
os.environ["JAVA_TOOL_OPTIONS"] = "-Djava.security.manager=allow"
```

```python

from pyspark.sql import SparkSession 
from pyspark.sql.functions import avg, col, max, desc, sum
```

```python
spark = SparkSession.builder.master(host).appName(app_name).getOrCreate()
```

```python
df = spark.read.parquet(data_file_path)
df = df.withColumn('Price', col('Price').cast('float'))
df = df.withColumn('Quantity', col('Quantity').cast('int'))
```
