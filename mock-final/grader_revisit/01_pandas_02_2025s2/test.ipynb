```python
df_trend = df[['trending_date','comment_count']].copy()
df_trend['vdo_count'] = 1
```

```python
df_trend['avg'] = df_trend['comment_count'].astype(int) / df_trend['vdo_count'].astype(int)
```

```python
df_trend['avg'].min()
```

```python
df_trend.loc[df_trend['avg'] == df_trend['avg'].min()]
```
