```python
class MushroomClassifier:
    def __init__(self, data_path): # DO NOT modify this line
        self.data_path = data_path
        self.df = pd.read_csv(data_path)

    def Q1(self): # DO NOT modify this line
        """
            1. (From step 1) Before doing the data prep., how many "na" are there in "gill-size" variables?
        """
        # remove pass and replace with you code
        return self.df["gill-size"].isna().sum()

    def Q2(self): # DO NOT modify this line
        """
            2. (From step 2-4) How many rows of data, how many variables?
            - Drop rows where the target (label) variable is missing.
            - Drop the following variables:
            'id','gill-attachment', 'gill-spacing', 'gill-size','gill-color-rate','stalk-root', 'stalk-surface-above-ring',
            'stalk-surface-below-ring', 'stalk-color-above-ring-rate','stalk-color-below-ring-rate','veil-color-rate','veil-type'
            - Examine the number of rows, the number of digits, and whether any are missing.
        """
        # Step 2
        df = self.df.copy()
        drop_cond = df["label"].isna()
        df = df[~drop_cond]

        # Step 3
        df = df.drop(
            labels=[
                "id",
                "gill-attachment",
                "gill-spacing",
                "gill-size",
                "gill-color-rate",
                "stalk-root",
                "stalk-surface-above-ring",
                "stalk-surface-below-ring",
                "stalk-color-above-ring-rate",
                "stalk-color-below-ring-rate",
                "veil-color-rate",
                "veil-type",
            ],
            axis=1
        )
        return df.shape

    def Q3(self): # DO NOT modify this line
        """
            3. (From step 5-6) Answer the quantity class0:class1
            - Fill missing values by adding the mean for numeric variables and the mode for nominal variables.
            - Convert the label variable e (edible) to 1 and p (poisonous) to 0 and check the quantity. class0: class1
            - Note: You need to reproduce the process (code) from Q2 to obtain the correct result.
        """
        # Step 2
        df = self.df.copy()
        df = df.dropna(subset=["label"])

        # Step 3
        df = df.drop(
            labels=[
                "id",
                "gill-attachment",
                "gill-spacing",
                "gill-size",
                "gill-color-rate",
                "stalk-root",
                "stalk-surface-above-ring",
                "stalk-surface-below-ring",
                "stalk-color-above-ring-rate",
                "stalk-color-below-ring-rate",
                "veil-color-rate",
                "veil-type",
            ],
            axis=1,
        )

        # Step 5
        df = df.fillna(df.mean(numeric_only=True)) # Fill numeric var
        df = df.fillna(df.select_dtypes(include='object').mode().iloc[0]) # fill mode for nominal var

        # Step 6
        df["label"] = df["label"].map({'e':1, 'p':0})
        return (sum(df["label"] == 0),sum(df["label"] == 1))

    def Q4(self): # DO NOT modify this line
        """
            4. (From step 7-8) How much is each training and testing sets
            - Convert the nominal variable to numeric using a dummy code with drop_first = True.
            - Split train/test with 20% test, stratify, and seed = 2020.
            - Note: You need to reproduce the process (code) from Q2, Q3 to obtain the correct result.
        """
        # Step 2
        df = self.df.copy()
        df = df.dropna(subset=["label"])

        # Step 3
        df = df.drop(
            labels=[
                "id",
                "gill-attachment",
                "gill-spacing",
                "gill-size",
                "gill-color-rate",
                "stalk-root",
                "stalk-surface-above-ring",
                "stalk-surface-below-ring",
                "stalk-color-above-ring-rate",
                "stalk-color-below-ring-rate",
                "veil-color-rate",
                "veil-type",
            ],
            axis=1,
        )

        # Step 5
        df = df.fillna(df.mean(numeric_only=True))  # Fill numeric var
        df = df.fillna(
            df.select_dtypes(include="object").mode().iloc[0]
        )  # fill mode for nominal var

        # Step 6
        df["label"] = df["label"].map({"e": 1, "p": 0})

        # Step 7
        nominal_var = df.select_dtypes("object").columns.values
        df = pd.get_dummies(df, columns=nominal_var, drop_first=True)

        # Step 8 Split train and test
        X = df.drop(labels=["label"], axis=1)
        y = df["label"]
        X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=2020, stratify=y)
        return (X_train.shape, X_test.shape)

    def Q5(self):
        """
            5. (From step 9) Best params after doing random forest grid search.
            Create a Random Forest with GridSearch on training data with 5 CV.
            - 'criterion':['gini','entropy']
            - 'max_depth': [2,3]
            - 'min_samples_leaf':[2,5]
            - 'N_estimators':[100]
            - 'random_state': 2020
            - Note: You need to reproduce the process (code) from Q2, Q3, Q4 to obtain the correct result.
        """
        # Step 2
        df = self.df.copy()
        df = df.dropna(subset=["label"])

        # Step 3
        df = df.drop(
            labels=[
                "id",
                "gill-attachment",
                "gill-spacing",
                "gill-size",
                "gill-color-rate",
                "stalk-root",
                "stalk-surface-above-ring",
                "stalk-surface-below-ring",
                "stalk-color-above-ring-rate",
                "stalk-color-below-ring-rate",
                "veil-color-rate",
                "veil-type",
            ],
            axis=1,
        )

        # Step 5
        df = df.fillna(df.mean(numeric_only=True))  # Fill numeric var
        df = df.fillna(
            df.select_dtypes(include="object").mode().iloc[0]
        )  # fill mode for nominal var

        # Step 6
        df["label"] = df["label"].map({"e": 1, "p": 0})

        # Step 7
        nominal_var = df.select_dtypes("object").columns.values
        df = pd.get_dummies(df, columns=nominal_var, drop_first=True)

        # Step 8 Split train and test
        X = df.drop(labels=["label"], axis=1)
        y = df["label"]
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.2, random_state=2020, stratify=y
        )

        # Step 9 ref: https://www.kaggle.com/code/sociopath00/random-forest-using-gridsearchcv
        params = {'criterion':['gini','entropy'],'max_depth': [2,3],'min_samples_leaf':[2,5],'n_estimators':[100]}
        rfc = RandomForestClassifier(random_state=2020)

        CV_rfc = GridSearchCV(estimator=rfc, param_grid=params, cv=5)
        CV_rfc.fit(X_train, y_train)
        best_param = CV_rfc.best_params_
        return (
            best_param["criterion"],
            best_param["max_depth"],
            best_param["min_samples_leaf"],
            best_param["n_estimators"],
            2020
        )

    def Q6(self):
        """
            5. (From step 10) What is the value of macro f1 (2 digits)?
            Predict the testing data set with confusion_matrix and classification_report,
            using scientific rounding (less than 0.5 dropped, more than 0.5 then increased)
            - Note: You need to reproduce the process (code) from Q2, Q3, Q4, Q5 to obtain the correct result.
        """
        # Step 2
        df = self.df.copy()
        df = df.dropna(subset=["label"])

        # Step 3
        df = df.drop(
            labels=[
                "id",
                "gill-attachment",
                "gill-spacing",
                "gill-size",
                "gill-color-rate",
                "stalk-root",
                "stalk-surface-above-ring",
                "stalk-surface-below-ring",
                "stalk-color-above-ring-rate",
                "stalk-color-below-ring-rate",
                "veil-color-rate",
                "veil-type",
            ],
            axis=1,
        )

        # Step 5
        df = df.fillna(df.mean(numeric_only=True))  # Fill numeric var
        df = df.fillna(
            df.select_dtypes(include="object").mode().iloc[0]
        )  # fill mode for nominal var

        # Step 6
        df["label"] = df["label"].map({"e": 1, "p": 0})

        # Step 7
        nominal_var = df.select_dtypes("object").columns.values
        df = pd.get_dummies(df, columns=nominal_var, drop_first=True)

        # Step 8 Split train and test
        X = df.drop(labels=["label"], axis=1)
        y = df["label"]
        X_train, X_test, y_train, y_test = train_test_split(
            X, y, test_size=0.2, random_state=2020, stratify=y
        )

        # Step 9 ref: https://www.kaggle.com/code/sociopath00/random-forest-using-gridsearchcv
        params = {'criterion':['gini','entropy'],'max_depth': [2,3],'min_samples_leaf':[2,5],'n_estimators':[100]}
        rfc = RandomForestClassifier(random_state=2020)

        CV_rfc = GridSearchCV(estimator=rfc, param_grid=params, cv=5)
        CV_rfc.fit(X_train, y_train)
        pred = CV_rfc.predict(X_test)
        report = classification_report(y_test, pred, output_dict=True)
        # print(classification_report(y_test, pred))
        # return round(report["macro avg"]["f1-score"] + 0.5, 2)
        return (round(report["0"]["f1-score"],2), round(report["1"]["f1-score"],2))
```
